```python
import google.generativeai as genai
import os

from dotenv import load_dotenv
load_dotenv()

# 🔹 Step 1: Set up your Gemini API Key
# os.environ["GOOGLE_API_KEY"] = "your-api-key-here"  # 🔹 Replace with your API Key
genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))

# 🔹 Step 2: Load Gemini Model
model = genai.GenerativeModel("models/gemini-2.0-flash")

# 🔹 Step 3: Few-shot Prompt for SATD Detection
few_shot_prompt = """
You are an AI trained to detect Self-Admitted Technical Debt (SATD) in software comments.
Classify each sentence as "SATD" (Self-Admitted Technical Debt) or "Not SATD".

Example 1:
"This code is a mess. I should refactor it." → SATD

Example 2:
"Refactored the function to improve readability." → Not SATD

Example 3:
"I know this is a hack, but it works." → SATD

Example 4:
"Fixed a bug in the authentication system." → Not SATD

Now classify this sentence:
"""

# 🔹 Step 4: Function to Classify SATD
def classify_satd(sentence):
    """Classifies SATD using Google Gemini AI."""

    prompt = few_shot_prompt + f'"{sentence}" →'

    # Generate response
    response = model.generate_content(prompt)

    # Extract prediction
    return response.text.strip()

# 🔹 Step 5: Interactive Testing
if __name__ == "__main__":
    print("\n🔹 Self-Admitted Technical Debt (SATD) Classifier using Google Gemini")
    print("Type a sentence to classify (or type 'exit' to quit)\n")

    while True:
        test_sentence = input("Enter a sentence: ").strip()
        if test_sentence.lower() == "exit":
            break

        label = classify_satd(test_sentence)
        print(f"➡️ Predicted Label: {label}\n")
```

```python
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer

# Load BERT model & tokenizer for classification
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"  # Pretrained classification model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Move model to CPU
device = torch.device("cpu")
model.to(device)

# Few-shot examples
few_shot_examples = [
    ("This code is a mess. I should refactor it.", "SATD"),
    ("Refactored the function to improve readability.", "Not SATD"),
    ("I know this is a hack, but it works.", "SATD"),
    ("Fixed a bug in the authentication system.", "Not SATD")
]


# Function to classify SATD using BERT
def classify_satd(sentence):
    """Classifies SATD using BERT."""

    # Tokenize input
    inputs = tokenizer(sentence, return_tensors="pt", truncation=True, padding=True)
    inputs = {key: value.to(device) for key, value in inputs.items()}  # Move to CPU

    # Run inference
    with torch.no_grad():
        outputs = model(**inputs)

    # Get prediction
    logits = outputs.logits
    predicted_class = torch.argmax(logits, dim=1).item()

    # Map prediction to label (Adjust based on model)
    label_map = {0: "Not SATD", 1: "SATD"}
    return label_map.get(predicted_class, "Unknown")


# Interactive Testing
if __name__ == "__main__":
    print("\n🔹 Self-Admitted Technical Debt (SATD) Classifier using BERT")
    print("Type a sentence to classify (or type 'exit' to quit)\n")

    while True:
        test_sentence = input("Enter a sentence: ").strip()
        if test_sentence.lower() == "exit":
            break

        label = classify_satd(test_sentence)
        print(f"➡️ Predicted Label: {label}\n")
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Load GPT-2 tokenizer and model
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

# Define your SATD classification function
def classify_satd(text):
    prompt = f"Is the following text related to technical debt? Text: {text} Answer:"

    # Encode the prompt with padding and attention mask
    inputs = tokenizer(prompt, return_tensors="pt", padding=True, truncation=True, max_length=512)

    # Ensure pad_token_id is set
    inputs["pad_token_id"] = tokenizer.eos_token_id  # Using eos_token_id as pad_token_id

    # Generate response
    outputs = model.generate(
        inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_length=100,
        num_return_sequences=1,
        pad_token_id=tokenizer.eos_token_id  # Explicitly setting pad_token_id for generation
    )

    # Decode the output and return the classification
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response

# Test the classification function
text = "This code contains a temporary solution to avoid a bigger problem."
result = classify_satd(text)
print(result)  # Example: "Yes, this is related to technical debt."
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Load the GPT-2 model and tokenizer
model = AutoModelForCausalLM.from_pretrained("gpt2",  torch_dtype=torch.float16, attn_implementation="flash_attention_2")
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Function to classify SATD using GPT-2
def detect_satd(prompt):
    # Prefix for SATD detection task
    satd_detection_prompt = "Is this statement a Self-Admitted Technical Debt? '"
    prompt_text = satd_detection_prompt + prompt + "'"

    # Tokenize the input text
    input_ids = tokenizer(prompt_text, return_tensors="pt").input_ids

    # Generate text with sampling for prediction
    gen_tokens = model.generate(
        input_ids,
        do_sample=True,
        temperature=0.9,
        max_length=100,
        top_p=0.95,
        top_k=60
    )

    # Decode the generated tokens into text
    gen_text = tokenizer.batch_decode(gen_tokens, skip_special_tokens=True)[0]

    # Extract the prediction from the generated text (you may refine this to better handle the output)
    if "yes" in gen_text.lower():
        return "SATD Detected"
    else:
        return "No SATD Detected"

# Example usage
example_prompt = "The function is not optimized and can be improved later."
result = detect_satd(example_prompt)
print(result)
```

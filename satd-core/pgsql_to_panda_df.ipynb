```python
import pandas as pd
from dotenv import load_dotenv

from comment import CommentRepository
from db_config import SessionLocal

load_dotenv()
session = SessionLocal()
repo = CommentRepository()
comments = repo.list_all_comments()
df = pd.DataFrame(columns=[
    "id",
    "text",
    "is_td",
    "gemini_td",
    "gpt4_td",
    "uid",
    "repository_id",
    "repository_directory",
    "start_line",
    "end_line",
    "file",
    "comment_hash",
    "commit_hash",
    "is_random",
    "note",
    "language",
    "created_at",
    "updated_at"
]
)
ids = []
texts = []
labels = []
for comment in comments:
    row = {
        "id": comment.id,
        "text": comment.text,
        "is_td": comment.is_td,
        "gemini_td": None,
        "gpt4_td": None,
        "uid": comment.uid,
        "repository_id": comment.repository_id,
        "repository_directory": comment.repository_directory,
        "start_line": comment.start_line,
        "end_line": comment.end_line,
        "file": comment.file,
        "comment_hash": comment.commit_hash,
        "commit_hash": comment.commit_hash,
        "is_random": comment.is_random,
        "note": comment.note,
        "language": comment.language,
        "created_at": comment.created_at,
        "updated_at": comment.updated_at,
    }
    df = pd.concat([df, pd.DataFrame([row])], ignore_index=True)
df.to_csv('../data/comments.csv', index=False)
```

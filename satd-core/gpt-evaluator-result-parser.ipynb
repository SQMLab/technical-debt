```python
import collections

import pandas as pd
from LlmOutputLabelConverter import LlmOutputLabelConverter
from sklearn.metrics import classification_report


import json
import os
output_label_converter = LlmOutputLabelConverter({'yes', 'no'}, 'no')
result_map = collections.defaultdict(list)
with open('../cache/output/eval_items_OutputDataItemStatusParam.ALL_2025-09-20_22-27-06.jsonl', 'r') as f:
    for line in f:
        output = json.loads(line)
        run_id = output['run_id']
        item = output['item']
        predicted_output_raw = output['sample']['outputs'][-1]['content']
        item['label_pred']  = output_label_converter.convert_label(predicted_output_raw)
        item['label_pred_raw'] = predicted_output_raw
        result_map[run_id].append(item)
for run_id, items in result_map.items():
    print(f'{run_id} {len(items)}')
    df = pd.DataFrame(items)
    os.makedirs('../cache/output/evaluation', exist_ok=True)
    mdf = df.copy()
    df.to_csv(f'../cache/output/evaluation/{run_id}.csv', index=False)
    print(classification_report(df['label'], df['label_pred'], zero_division=0, digits=3, output_dict=False))
    mdf = df[~df['label_pred_raw'].isna()]
    mdf = mdf[mdf['label'] != mdf['label_pred']]
    mdf.to_csv(f'../cache/output/evaluation/{run_id}_mismatched.csv', index=False)



```

```python
import os
import pandas as pd
from pathlib import Path
from tqdm import tqdm

merged_dir = Path("../cache/output/merged")
# Get all CSV files in the merged directoryOOO
csv_files = list(merged_dir.glob("*.csv"))

for test_name, test_file in ['unique', '../data/unique_detect_test.csv'], ['duplicate',
                                                                           '../data/duplicate_detect_test.csv']:
    comment_ids = set(pd.read_csv(test_file)["id"].tolist())

    for file_path in csv_files:
        try:
            # Read the merged result file
            # We use low_memory=False in case there are mixed types in large files
            df = pd.read_csv(file_path, low_memory=False)

            # Filter for unique
            df = df[df['id'].isin(comment_ids)]
            new_file_name = file_path.name.replace("merged_", "")
            print(f"{new_file_name}: {len(df)}")
            out_path = f'../cache/output/detect/{test_name}/{new_file_name}'
            os.makedirs(os.path.dirname(out_path), exist_ok=True)
            df.to_csv(out_path, index=False)
        except pd.errors.EmptyDataError:
            print(f"\nWarning: {file_path.name} is empty, skipping.")
        except Exception as e:
            print(f"\nError processing {file_path.name}: {e}")

print("Done!")
```
